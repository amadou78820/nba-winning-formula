```python
def keep_season_total(group):

    # Plusieurs lignes + présence de TOT
    if "TOT" in group["TEAM_ABBREVIATION"].values:
        return group[
            group["TEAM_ABBREVIATION"] == "TOT"
        ].iloc[0]

    # Une seule ligne
    return group.iloc[0]


df_player_season = (
    df_goat
    .groupby(
        ["PLAYER_ID", "SEASON_ID"],
        group_keys=False
    )
    .apply(keep_season_total)
    .reset_index(drop=True)
)

print("Avant :", df_goat.shape)
print("Après :", df_player_season.shape)
```

```python
check = (
    df_player_season
    .groupby(["PLAYER_ID", "SEASON_ID"])
    .size()
)

print(
    "Nombre de doublons restants :",
    (check > 1).sum()
)
```

```python
goat_players = [
    "Michael Jordan",
    "LeBron James",
    "Kareem Abdul-Jabbar",
    "Magic Johnson",
    "Larry Bird"
]

career_check = (
    df_player_season[
        df_player_season["Name"].isin(goat_players)
    ]
    .groupby("Name")
    .agg(
        first_season=("SEASON_ID", "min"),
        last_season=("SEASON_ID", "max"),
        seasons=("SEASON_ID", "nunique"),
        games=("GP", "sum"),
        points=("PTS", "sum"),
        rebounds=("REB", "sum"),
        assists=("AST", "sum")
    )
    .sort_values("points", ascending=False)
)

display(career_check)
```

```python
df_player_season["PPG"] = (
    df_player_season["PTS"] / df_player_season["GP"]
)

df_player_season["RPG"] = (
    df_player_season["REB"] / df_player_season["GP"]
)

df_player_season["APG"] = (
    df_player_season["AST"] / df_player_season["GP"]
)

df_player_season["SPG"] = (
    df_player_season["STL"] / df_player_season["GP"]
)

df_player_season["BPG"] = (
    df_player_season["BLK"] / df_player_season["GP"]
)

df_player_season["MPG"] = (
    df_player_season["MIN"] / df_player_season["GP"]
)
```

```python
stats_per_game = [
    "PPG",
    "RPG",
    "APG",
    "SPG",
    "BPG",
    "MPG"
]

df_player_season[stats_per_game] = (
    df_player_season[stats_per_game]
    .round(2)
)
```

```python
goat_seasons = df_player_season[
    df_player_season["Name"].isin(goat_players)
].copy()

display(
    goat_seasons[
        [
            "Name",
            "SEASON_ID",
            "PLAYER_AGE",
            "GP",
            "PPG",
            "RPG",
            "APG",
            "SPG",
            "BPG",
            "FG_PCT",
            "FG3_PCT",
            "FT_PCT"
        ]
    ]
    .sort_values(["Name", "SEASON_ID"])
)
```

```python
peak_ppg = (
    goat_seasons
    .sort_values(
        ["Name", "PPG"],
        ascending=[True, False]
    )
    .groupby("Name")
    .head(5)
)

display(
    peak_ppg[
        [
            "Name",
            "SEASON_ID",
            "PLAYER_AGE",
            "GP",
            "PPG",
            "RPG",
            "APG",
            "FG_PCT"
        ]
    ]
    .sort_values(
        ["Name", "PPG"],
        ascending=[True, False]
    )
)
```

```python
peak_5 = (
    peak_ppg
    .groupby("Name")
    .agg(
        peak_5_ppg=("PPG", "mean"),
        peak_5_rpg=("RPG", "mean"),
        peak_5_apg=("APG", "mean")
    )
    .round(2)
    .sort_values("peak_5_ppg", ascending=False)
)

display(peak_5)
```

```python
career_stats = (
    df_player_season[
        df_player_season["Name"].isin(goat_players)
    ]
    .groupby("Name")
    .agg(
        seasons=("SEASON_ID", "nunique"),
        games=("GP", "sum"),
        minutes=("MIN", "sum"),
        points=("PTS", "sum"),
        rebounds=("REB", "sum"),
        assists=("AST", "sum"),
        steals=("STL", "sum"),
        blocks=("BLK", "sum")
    )
)
```

```python
career_stats["career_ppg"] = (
    career_stats["points"] / career_stats["games"]
)

career_stats["career_rpg"] = (
    career_stats["rebounds"] / career_stats["games"]
)

career_stats["career_apg"] = (
    career_stats["assists"] / career_stats["games"]
)

career_stats["career_spg"] = (
    career_stats["steals"] / career_stats["games"]
)

career_stats["career_bpg"] = (
    career_stats["blocks"] / career_stats["games"]
)

career_stats["career_mpg"] = (
    career_stats["minutes"] / career_stats["games"]
)
```

```python
def minmax_100(series):
    min_val = series.min()
    max_val = series.max()

    if max_val == min_val:
        return pd.Series(
            50,
            index=series.index
        )

    return (
        (series - min_val)
        / (max_val - min_val)
        * 100
    )
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_absolute_error
import pandas as pd

features = [
    "avg_fg_pct",
    "avg_fg3_pct",
    "avg_ft_pct",
    "avg_dreb",
    "avg_ast",
    "avg_stl",
    "avg_blk",
    "avg_tov"
]

def temporal_model(start_year):

    train = team_season[
        (team_season["season_year"] >= start_year)
        &
        (team_season["season_year"] <= 2017)
    ].copy()

    test = team_season[
        team_season["season_year"] >= 2018
    ].copy()

    train_data = train[
        features + ["win_rate"]
    ].dropna()

    test_data = test[
        features + ["win_rate"]
    ].dropna()

    scaler = StandardScaler()

    X_train = scaler.fit_transform(
        train_data[features]
    )

    X_test = scaler.transform(
        test_data[features]
    )

    y_train = train_data["win_rate"]
    y_test = test_data["win_rate"]

    model = LinearRegression()

    model.fit(
        X_train,
        y_train
    )

    train_pred = model.predict(X_train)
    test_pred = model.predict(X_test)

    return {
        "start_year": start_year,
        "train_obs": len(train_data),

        "train_r2": r2_score(
            y_train,
            train_pred
        ),

        "test_r2": r2_score(
            y_test,
            test_pred
        ),

        "test_mae": mean_absolute_error(
            y_test,
            test_pred
        )
    }
```

```python
results = []

for start_year in [
    team_season["season_year"].min(),
    1990,
    2000,
    2010
]:
    results.append(
        temporal_model(start_year)
    )

window_results = pd.DataFrame(results)

window_results["approx_win_error"] = (
    window_results["test_mae"] * 82
)

display(
    window_results.round(3)
)
```

```python
from sklearn.preprocessing import StandardScaler
import numpy as np
```

```python
features = [
    "PRODUCTION_Z",
    "TS_PLUS",
    "Recognition_Trajectory"
]

reference = pd.concat([
    future_model[features],
    goat_trajectory[features]
]).dropna()

scaler_similarity = StandardScaler()

scaler_similarity.fit(reference)
```

```python
similarity_rows = []

for _, candidate in future_model.iterrows():

    age = candidate["AGE"]

    goats_same_age = goat_trajectory[
        goat_trajectory["AGE"] == age
    ]

    if goats_same_age.empty:
        continue

    candidate_vector = scaler_similarity.transform(
        pd.DataFrame(
            [candidate[features].values],
            columns=features
        )
    )[0]

    for _, goat in goats_same_age.iterrows():

        goat_vector = scaler_similarity.transform(
            pd.DataFrame(
                [goat[features].values],
                columns=features
            )
        )[0]

        distance = np.linalg.norm(
            candidate_vector - goat_vector
        )

        similarity = 100 / (1 + distance)

        similarity_rows.append({
            "Name": candidate["Name"],
            "AGE": age,
            "GOAT": goat["Name"],
            "Distance": distance,
            "Similarity": similarity
        })

goat_similarity = pd.DataFrame(similarity_rows)
```

```python
closest_goat = (
    goat_similarity
    .sort_values(
        ["Name", "Similarity"],
        ascending=[True, False]
    )
    .groupby("Name")
    .first()
    .reset_index()
)
```
